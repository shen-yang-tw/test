```python
# 參數(parameter)是函式的輸入值，引數(argument) --> 是呼叫函式時傳遞給參數的值。
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
```

```python
# 引數位置 (positional argument) --> 依照參數的順序傳遞給函式。
def pet(animal, name):
    print(f"我養了一隻{animal}，它叫{name}。")

pet("狗", "小白")  # 正確的引數位置

# 引數名稱 (keyword argument) --> 透過指定參數名稱來傳遞引數。
pet(animal="貓", name="小黑")  # 使用引數名稱傳遞引數
pet(name="小黑", animal="貓")  # 可以改變引數的順序，因為使用了引數名稱
```

```python
## 重要規則：混用時 "位置引數" 必須放在 "關鍵字引數" 前面，否則會出現語法錯誤。
pet("狗", name="小白")  # 正確的混用方式
```

```python
##  搭配預設值 (Default Arguments) 使用，如果參數有預設值，呼叫函式時可以選擇性地傳遞引數，若未傳遞則使用預設值。
def order_drink(drink, ice="正常冰", sugar="半糖"):
    print(f"點餐：{drink} {ice} / {sugar}")

order_drink("珍珠奶茶")  # 使用預設值
order_drink("綠茶", ice="少冰")  # 指定冰量
order_drink("紅茶", sugar="無糖")  # 指定糖分
```

```python
## 特殊符號 / 與 * 的使用：/ 表示前面的參數只能使用位置引數傳遞，* 表示後面的參數只能使用關鍵字引數傳遞。
def calc_bmi(weight, height):
    bmi = weight / (height ** 2)
    print(f"體重：{weight} 公斤，身高：{height} 公尺，BMI：{bmi:.2f}")

calc_bmi(70, 1.75)  # 調用函式並傳遞引數
```

```python
## def 函式名稱(僅有位置引數的參數 , / , 位置或關鍵字引數皆可 , * , 僅有關鍵字引數的參數):
def register(id, /, name, *, email):
    print(f"註冊成功！ID: {id}, 姓名: {name}, 電子郵件: {email}")

register(123, "Alice", email="alice@example.com")
```

```python
## 沒有限定參數的函式，使用 *args 來接收任意數量的位置引數，使用 **kwargs 來接收任意數量的關鍵字引數。
def sum_squares(*numbers):
    total = sum(x ** 2 for x in numbers)
    return total

result = sum_squares(1, 2, 3, 4)
print(f"平方和: {result}")
```

```python
## 使用 **kwargs 來接收任意數量的關鍵字引數，並將其存儲在字典中。
def build_profile(name, age, **info):
    profile = {
        "name": name,
        "age": age
    }
    profile.update(info)
    return profile

user_profile = build_profile("Alice", 30, location="New York", profession="Engineer")
print(user_profile)
```
